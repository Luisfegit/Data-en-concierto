# %% [6] Diccionario breve de variables

```python

# Significado y rol son conocidos de antemano; tipo, nulos y únicos se calculan.
significado = {
    "track_id":           ("Identificador único de la canción en Spotify", "identificador"),
    "track_name":         ("Nombre de la canción", "texto"),
    "track_number":       ("Posición de la canción dentro de su álbum", "numérica"),
    "track_popularity":   ("Popularidad de la canción (0 a 100)", "resultado de interés"),
    "track_duration_ms":  ("Duración en milisegundos", "numérica"),
    "track_duration_min": ("Duración en minutos", "numérica"),
    "explicit":           ("Si la canción tiene contenido explícito", "categórica (booleana)"),
    "artist_name":        ("Nombre del artista principal", "categórica (alta cardinalidad)"),
    "artist_popularity":  ("Popularidad del artista (escala 0 a 1000)", "numérica / posible resultado"),
    "artist_followers":   ("Seguidores del artista en Spotify", "numérica"),
    "artist_genres":      ("Lista de géneros del artista, guardada como texto", "categórica (lista)"),
    "album_id":           ("Identificador único del álbum", "identificador"),
    "album_name":         ("Nombre del álbum", "texto"),
    "album_release_date": ("Fecha de lanzamiento del álbum", "fecha"),
    "album_total_tracks": ("Total de canciones del álbum", "numérica"),
    "album_type":         ("Tipo de lanzamiento: album, single o compilation", "categórica"),
}

def observaciones(col):
    """Observaciones calculadas a partir de los datos reales."""
    s = df[col]
    obs = []
    if s.isna().any():
        obs.append(f"{s.isna().sum()} nulos")
    if col == "artist_popularity":
        obs.append(f"rango {s.min():.0f}-{s.max():.0f}: escala x10 respecto a Spotify (0-100)")
    if col == "track_popularity":
        obs.append(f"{(s == 0).sum()} canciones con popularidad 0")
    if col == "track_duration_ms":
        obs.append(f"{(s == 0).sum()} con duración 0; promedio {s.mean()/60000:.1f} min")
    if col == "track_duration_min":
        obs.append(f"{(s == 0).sum()} con duración 0; promedio {s.mean():.1f} min")
    if col == "artist_genres":
        obs.append(f"{(s == '[]').sum()} listas vacías ({(s == '[]').mean()*100:.0f} %); hay que convertir a lista")
    if col == "album_release_date":
        obs.append(f"{(s.astype(str).str.len() == 4).sum()} fechas solo con año; convertir con to_datetime")
    if col == "explicit":
        obs.append(f"{s.mean()*100:.0f} % son explícitas")
    if col in ("track_id", "album_id"):
        obs.append("único por canción" if s.is_unique else "se repite (varias canciones por álbum)")
    return "; ".join(obs) if obs else "-"

diccionario = pd.DataFrame({
    "variable":      df.columns,
    "significado":   [significado[c][0] for c in df.columns],
    "tipo":          [str(df[c].dtype) for c in df.columns],
    "rol":           [significado[c][1] for c in df.columns],
    "observaciones": [observaciones(c) for c in df.columns],
})
display(diccionario)
```

```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style="whitegrid")
plt.rcParams.update({"figure.dpi": 100, "axes.titlesize": 12, "axes.labelsize": 11})

# Reemplazamos la lectura del archivo local inexistente por una copia del DataFrame 'raw' ya cargado
d = raw.copy()
P = "track_popularity"
d["fecha"] = pd.to_datetime(d["album_release_date"])
d["anio"] = d["fecha"].dt.year
d["genero"] = d["artist_genres"].fillna("").str.split(", ").str[0].replace("", np.nan)  # primer género del artista
d["Contenido"] = d["explicit"].map({True: "Explícita", False: "No explícita"})
d["Tipo"] = d["album_type"].map({"album": "Álbum", "single": "Sencillo", "compilation": "Compilación"})

print("Canciones:", len(d), "| Artistas:", d["artist_name"].nunique())
print("Lanzamientos de", d["fecha"].min().date(), "a", d["fecha"].max().date())
print("Canciones lanzadas en 2024-2025:", round(d["anio"].isin([2024, 2025]).mean()*100, 1), "%")
print("Canciones sin género registrado:", d["genero"].isna().sum(), f"({d['genero'].isna().mean():.1%})")
```

```python
fig, ax = plt.subplots(figsize=(8, 4.5))
sns.histplot(d[P], bins=50, color="#4C72B0", ax=ax)
ax.axvline(d[P].median(), color="crimson", ls="--", label=f"Mediana = {d[P].median():.0f}")
ax.set(title=f"Distribución de la popularidad de las canciones (n = {len(d):,})",
       xlabel="Popularidad de la canción (índice 0–100)", ylabel="Número de canciones")
ax.legend(); plt.tight_layout(); plt.show()
print(f"Media: {d[P].mean():.1f} | Mediana: {d[P].median():.0f}")
print(f"Popularidad = 0: {(d[P]==0).mean():.1%} | Popularidad >= 70: {(d[P]>=70).mean():.1%}")
```

**Interpretación**

- **Patrón:** la distribución está sesgada hacia la izquierda: la mayoría de las canciones se concentra en valores medios y altos (mediana 58, media 52.4), con una cola larga hacia valores bajos y un pico en 0.
- **Cuánto representa:** 5.9 % de las canciones tiene popularidad 0 y 27.7 % tiene 70 o más.
- **Grupo o período:** todas las 8,582 canciones del archivo, con álbumes lanzados entre 1952 y 2025 (la mayoría, recientes).
- **Qué no permite concluir:** no explica *por qué* unas canciones son más populares que otras. Además, el índice de Spotify refleja la actividad de escucha reciente, por lo que puede favorecer canciones nuevas y no mide la aceptación histórica.

```python
con_genero = d["genero"].notna().sum()
top = d["genero"].value_counts().head(10)
fig, ax = plt.subplots(figsize=(8, 5))
sns.barplot(x=top.values, y=top.index, color="#55A868", ax=ax)
for i, v in enumerate(top.values):
    ax.text(v, i, f" {v/con_genero:.1%}", va="center", fontsize=9)
ax.set(title="Géneros principales más frecuentes (solo canciones con género registrado)",
       xlabel="Número de canciones", ylabel="Género principal del artista")
plt.tight_layout(); plt.show()
print("Canciones con género:", con_genero, "| Los 10 géneros suman:", f"{top.sum()/con_genero:.1%}")
```

**Interpretación**

- **Patrón:** el catálogo está repartido entre muchos géneros; los más frecuentes son country (572 canciones), pop (516) y soundtrack (352).
- **Cuánto representa:** country equivale al 11.0 % y pop al 9.9 % de las canciones con género registrado; los diez géneros mostrados suman 48.9 % de ellas.
- **Grupo o período:** solo las 5,221 canciones con género (60.8 % del total); se usa el primer género que aparece en la lista de cada artista.
- **Qué no permite concluir:** que un género sea más aceptado, porque la gráfica cuenta canciones del archivo y no escuchas. Tampoco se sabe si el 39 % sin género tiene una distribución distinta.

```python
sub = d[d["genero"].isin(top.index)]
orden = sub.groupby("genero")[P].median().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(9, 5))
sns.boxplot(data=sub, y="genero", x=P, order=orden.index, color="#8172B3", fliersize=2, ax=ax)
ax.set(title="Popularidad de la canción según el género principal (10 géneros más frecuentes)",
       xlabel="Popularidad de la canción (índice 0–100)", ylabel="Género principal del artista")
plt.tight_layout(); plt.show()
print(sub.groupby("genero")[P].agg(mediana="median", media="mean", n="count").loc[orden.index].round(1))
```

**Interpretación**

- **Patrón:** alternative pop (mediana 68), pop rock (67.5), rap (67) y pop (66) están en la parte alta; soundtrack (43) y dark R&B (52) están en la parte baja. Soundtrack también tiene la media más baja (37.2).
- **Cuánto representa:** entre alternative pop y soundtrack hay 25 puntos de diferencia en la mediana, y los cuatro géneros más altos superan en 8 puntos o más a la mediana global (58).
- **Grupo o período:** los 10 géneros más frecuentes de la gráfica 2 (2,555 canciones), de todas las épocas del archivo.
- **Qué no permite concluir:** causalidad. Las bandas sonoras suelen incluir piezas poco buscadas de forma individual, y la diferencia puede deberse a cómo se publican y no al gusto del público. Con un solo género por artista se pierde información de quienes mezclan géneros, y los géneros fuera del top 10 no están representados.

```python
rho = d[[P, "artist_popularity"]].corr(method="spearman").iloc[0, 1]
rho_f = d[[P, "artist_followers"]].corr(method="spearman").iloc[0, 1]
fig, ax = plt.subplots(figsize=(7, 5.5))
hb = ax.hexbin(d["artist_popularity"], d[P], gridsize=40, cmap="viridis", mincnt=1, bins="log")
plt.colorbar(hb, label="Número de canciones (escala logarítmica)")
ax.set(title=f"Popularidad del artista vs. popularidad de la canción (Spearman = {rho:.2f})",
       xlabel="Popularidad del artista (índice 0–100)", ylabel="Popularidad de la canción (índice 0–100)")
plt.tight_layout(); plt.show()
print(f"Spearman con popularidad del artista: {rho:.2f} | con seguidores del artista: {rho_f:.2f}")
```

**Interpretación**

- **Patrón:** relación positiva moderada: las canciones de artistas más populares tienden a ser más populares (Spearman = 0.47). Con el número de seguidores la asociación es parecida (0.42). También hay mucha dispersión: artistas muy populares tienen canciones poco escuchadas.
- **Cuánto representa:** la fama del artista explica una parte importante, pero no mayoritaria, de la variación; la correlación queda lejos de 1.
- **Grupo o período:** las 8,582 canciones, mezclando géneros y épocas.
- **Qué no permite concluir:** si el artista hace popular a la canción o si las canciones exitosas vuelven popular al artista (el sentido causal puede ir en ambas direcciones). Un mismo artista aparece en muchas canciones, así que las observaciones no son independientes.

```python
dd = d[d["track_duration_min"].between(1, 8)].copy()
dd["intervalo"] = pd.cut(dd["track_duration_min"], [1, 2, 2.5, 3, 3.5, 4, 5, 8], right=False)
t = dd.groupby("intervalo", observed=True)[P].agg(med="median", q1=lambda s: s.quantile(.25),
                                                    q3=lambda s: s.quantile(.75), n="count")
fig, ax = plt.subplots(figsize=(8.5, 4.8)); x = range(len(t))
ax.bar(x, t["med"], color="#DD8452")
ax.errorbar(x, t["med"], yerr=[t["med"]-t["q1"], t["q3"]-t["med"]], fmt="none", ecolor="k", capsize=3)
ax.set_xticks(list(x)); ax.set_xticklabels([f"{i.left:g}–{i.right:g}\n(n={n})" for i, n in zip(t.index, t["n"])])
ax.set(title="Popularidad mediana según la duración de la canción (barras de error = rango intercuartílico)",
       xlabel="Duración de la canción (minutos)", ylabel="Popularidad mediana (índice 0–100)")
plt.tight_layout(); plt.show()
rho_d = d[[P, "track_duration_min"]].corr(method="spearman").iloc[0, 1]
print(t.round(1)); print("Canciones fuera de 1-8 min:", len(d)-len(dd), "| Spearman global:", round(rho_d, 2))
print("Proporción entre 3 y 5 min:", f"{dd['track_duration_min'].between(3, 5, inclusive='left').mean():.1%}")
```

**Interpretación**

- **Patrón:** relación en forma de "U invertida": las canciones de 3 a 5 minutos tienen las medianas más altas (61–62); las más cortas (menos de 2 min, mediana 42) y las muy largas (5–8 min, mediana 56) quedan por debajo.
- **Cuánto representa:** cerca de dos tercios de las canciones entre 1 y 8 minutos dura entre 3 y 5 minutos. La asociación lineal global es débil (Spearman = 0.14), lo que indica que existe un rango de duración favorable y no una tendencia constante.
- **Grupo o período:** 8,494 canciones de entre 1 y 8 minutos (se excluyen 88 casos extremos), de todas las épocas.
- **Qué no permite concluir:** que alargar o acortar una canción la haga más popular. Las piezas muy cortas pueden ser intros, interludios o pistas auxiliares, que ya son menos escuchadas por su naturaleza.

```python
fig, ax = plt.subplots(figsize=(8.5, 4.8))
sns.boxplot(data=d, x="Tipo", y=P, hue="Contenido", order=["Álbum", "Sencillo", "Compilación"],
            palette=["#C44E52", "#4C72B0"], fliersize=2, ax=ax)
ax.set(title="Popularidad de la canción según el tipo de lanzamiento y el contenido explícito",
       xlabel="Tipo de lanzamiento", ylabel="Popularidad de la canción (índice 0–100)")
plt.tight_layout(); plt.show()
print(d.groupby(["Tipo", "Contenido"])[P].agg(mediana="median", n="count").round(1))
print("Medianas por contenido:", d.groupby("Contenido")[P].median().to_dict())
print("Medianas por tipo:", d.groupby("Tipo")[P].median().to_dict())
print("Proporción de canciones explícitas:", f"{d['explicit'].mean():.1%}")
```

**Interpretación**

- **Patrón:** las canciones explícitas tienen mayor mediana que las no explícitas (64 vs. 55) y la ventaja se repite en los tres tipos de lanzamiento (álbum: 66 vs. 59; sencillo: 55 vs. 50; compilación: 58 vs. 43). Las canciones de álbum son más populares (mediana 61) que los sencillos (51) y las compilaciones (44).
- **Cuánto representa:** 25.0 % de las canciones es explícita. Las compilaciones explícitas son solo 37 canciones, así que esa comparación es poco fiable.
- **Grupo o período:** las 8,582 canciones, separadas por tipo de lanzamiento y contenido.
- **Qué no permite concluir:** que ser explícita *cause* mayor popularidad. El contenido explícito está muy ligado a ciertos géneros (como el rap) y a artistas grandes, y no se controló por género ni por artista. Los sencillos suelen ser lanzamientos recientes que aún no acumulan reproducciones.
